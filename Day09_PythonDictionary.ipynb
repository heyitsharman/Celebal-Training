**CODIN CONCEPTS**

> **1: Standard IO (stdio) -** stdio stands for Standard Input and Output.

- The magic of f-strings is that you can embed Python expressions directly inside them by using curly braces `{}`.

````python
# PROG 1: School Result

# Write The Code Here

# SCHOOL RESULT USING DICTIONARY

import random

student_names = [f"Student{i}" for i in range(1, 21)]

students = {}

for name in student_names:
    students[name] = {
        "Gender": random.choice(["M", "F"]),
        "Physics Marks": round(random.uniform(0.0, 50.0), 1),
        "Chemistry Marks": round(random.uniform(0.0, 50.0), 1),
        "Maths Marks": round(random.uniform(0.0, 50.0), 1),
        "Days Attended": random.randint(1, 120)
    }


def calculate_result(student):
    total = (
        student["Physics Marks"]
        + student["Chemistry Marks"]
        + student["Maths Marks"]
    )

    percentage = (total / 150) * 100
    attendance_percentage = (student["Days Attended"] / 120) * 100

    if percentage > 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    elif percentage >= 30:
        grade = "D"
    else:
        grade = "F"

    student["Total Marks"] = round(total, 1)
    student["Percentage"] = round(percentage, 2)
    student["Attendance Percentage"] = round(attendance_percentage, 2)
    student["Grade"] = grade
    student["Result"] = "Pass" if percentage >= 30 else "Fail"

for name in students:
    calculate_result(students[name])


def display_result(name):
    student = students[name]

    print("\n" + "=" * 40)
    print("Name:", name)
    print("Gender:", student["Gender"])

    print("Marks:")
    print("- Physics:", student["Physics Marks"])
    print("- Chemistry:", student["Chemistry Marks"])
    print("- Maths:", student["Maths Marks"])

    print("Days Attended:", student["Days Attended"])
    print("Attendance Percentage:",
          str(student["Attendance Percentage"]) + "%")

    print(f"Total Marks: {student['Total Marks']}/150")
    print(f"Percentage Marks: {student['Percentage']}%")
    print("Grade:", student["Grade"])

    if student["Result"] == "Fail":
        print("Remarks: Failed, work hard to do better next time.")
    else:
        print("Remarks: Congratulations!, Passed Successfully.")


def display_all_results():
    choice = input(
        "\nEnter 'grade' to display results by grade "
        "or 'percentage' to display results by percentage: "
    ).strip().lower()

    if choice == "percentage":
        ordered_names = sorted(
            students,
            key=lambda name: students[name]["Percentage"],
            reverse=True
        )

    elif choice == "grade":
        grade_order = {
            "A+": 6, "A": 5, "B": 4,
            "C": 3, "D": 2, "F": 1
        }

        ordered_names = sorted(
            students,
            key=lambda name: (
                -grade_order[students[name]["Grade"]],
                -students[name]["Percentage"]
            )
        )

    else:
        print("Invalid choice! Please enter grade or percentage.")
        return

    for name in ordered_names:
        display_result(name)

def generate_summary():
    total_students = len(students)

    passed_students = [
        name for name in students
        if students[name]["Result"] == "Pass"
    ]

    pass_percentage = (len(passed_students) / total_students) * 100

    print("\n" + "=" * 50)
    print("              SCHOOL SUMMARY")
    print("=" * 50)

    print(f"Total Students: {total_students}")
    print(f"Students Passed: {len(passed_students)}")
    print(f"Students Failed: {total_students - len(passed_students)}")
    print(f"Overall Pass Percentage: {pass_percentage:.2f}%")
    print("\nGrade-wise Percentage Distribution:")

    for grade in ["A+", "A", "B", "C", "D", "F"]:
        count = sum(
            1 for student in students.values()
            if student["Grade"] == grade
        )
        distribution = (count / total_students) * 100
        print(f"Grade {grade}: {distribution:.2f}% ({count} students)")

    highest_total = max(
        student["Total Marks"] for student in students.values()
    )

    toppers = [
        name for name in students
        if students[name]["Total Marks"] == highest_total
    ]

    print("\nOverall Topper(s):", ", ".join(toppers))
    print("Highest Marks:", highest_total, "/150")

    print("\nSubject-wise Topper(s):")

    for subject in ["Physics Marks", "Chemistry Marks", "Maths Marks"]:
        highest = max(
            student[subject] for student in students.values()
        )

        subject_toppers = [
            name for name in students
            if students[name][subject] == highest
        ]

        print(
            f"{subject}: {', '.join(subject_toppers)} "
            f"({highest}/50)"
        )

    alphabetical_passed = sorted(passed_students)

    print("\nPassed Students (Alphabetical Order):")

    for name in alphabetical_passed:
        student = students[name]
        print(
            f"{name}: {student['Percentage']}%, "
            f"Grade {student['Grade']}"
        )

    marks_sorted_passed = sorted(
        passed_students,
        key=lambda name: students[name]["Total Marks"],
        reverse=True
    )

    print("\nPassed Students (Highest Marks First):")

    for name in marks_sorted_passed:
        student = students[name]
        print(
            f"{name}: {student['Total Marks']}/150, "
            f"{student['Percentage']}%, Grade {student['Grade']}"
        )


display_all_results()
generate_summary()

````
